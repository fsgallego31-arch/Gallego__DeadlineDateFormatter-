```python
# Activity 2: Deadline & Date Formatter
# Input Data from user
from datetime import datetime

# Collect user inputs for the event name and target date components
event_title = str(input("Enter the name of the event: "))
year =  int(input("Enter the year of the event (YYYY): "))
month = int(input("Enter the month of the event (1-12): "))
day = int(input("Enter the day of the event (1-31): "))

# Cleans and standardizes the event title
clean_event_title = event_title.strip().upper() #cleans the event title.

# A datetime object from the integer inputs to extract calendar details
event_date = datetime(year, month, day)
month_name = event_date.strftime("%B")
weekday_name = event_date.strftime("%A")

# Format the final date string
formatted_date = f"{month_name} {day}, {year}" #formats the date.

print("\n--- Formatted Output ---")
print(f"Event: {clean_event_title}")
print(f"Date: {formatted_date}")
print(f"Year: {year}")
print(f"Month: {month_name}")
print(f"Day: {day}")
print(f"Weekday: {weekday_name}")

```
